```python
from typing import List

class Solution:
    def kidsWithCandies(self, candies: List[int], extraCandies: int) -> List[bool]:
        # Find the maximum candies any kid currently has
        max_candies = max(candies)
        
        # For each kid, check if giving them extraCandies makes them >= max_candies
        result = [(candy + extraCandies) >= max_candies for candy in candies]
        
        return result

# Testing the Solution class for LeetCode 1431

solution = Solution()

test_cases = [
    {"candies": [2,3,5,1,3], "extraCandies": 3},
    {"candies": [4,2,1,1,2], "extraCandies": 1},
    {"candies": [12,1,12], "extraCandies": 10},
    {"candies": [5,5,5,5], "extraCandies": 2},   # All equal case
]

for case in test_cases:
    candies, extraCandies = case["candies"], case["extraCandies"]
    print(f"Input: candies={candies}, extraCandies={extraCandies}")
    print(f"Output: {solution.kidsWithCandies(candies, extraCandies)}")
    print("-" * 40)
```
