```python
from typing import List

class Solution:
    def shuffle(self, nums: List[int], n: int) -> List[int]:
        result = []
        
        # First half: x1, x2, ..., xn
        # Second half: y1, y2, ..., yn
        for i in range(n):
            result.append(nums[i])      # xi
            result.append(nums[i + n])  # yi
        
        return result


# Testing the Solution class for LeetCode 1470: Shuffle the Array

solution = Solution()

# Define test cases
test_cases = [
    {"nums": [2,5,1,3,4,7], "n": 3},       # Example 1
    {"nums": [1,2,3,4,4,3,2,1], "n": 4},   # Example 2
    {"nums": [1,1,2,2], "n": 2},           # Example 3
    {"nums": [10,20,30,40,50,60], "n": 3}, # Custom case
]

# Run and print results
for case in test_cases:
    nums, n = case["nums"], case["n"]
    print(f"Input: nums={nums}, n={n}")
    print(f"Output: {solution.shuffle(nums, n)}")
    print("-" * 40)
```
