```python
from typing import List

class Solution:
    def runningSum(self, nums: List[int]) -> List[int]:
        # Initialize result list
        running_sum = []
        current_sum = 0
        
        # Iterate through nums and build running sum
        for num in nums:
            current_sum += num
            running_sum.append(current_sum)
        
        return running_sum

# Testing the Solution class
nums_list = [
    [1, 2, 3, 4],        # Example 1
    [1, 1, 1, 1, 1],     # Example 2
    [3, 1, 2, 10, 1],    # Example 3
    [5],                 # Edge case: single element
    [0, 0, 0, 0],        # Edge case: all zeros
]

solution = Solution()

for nums in nums_list:
    print(f"Input: {nums}")
    print(f"Output: {solution.runningSum(nums)}")
    print("-" * 30)
```
